```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('USA_Housing.csv')
df.head()
```

```python
#1 Using the scatter Plot
features = [
    'Avg. Area Income',
    'Avg. Area House Age',
    'Avg. Area Number of Rooms',
    'Avg. Area Number of Bedrooms',
    'Area Population'
]

fig, axes = plt.subplots(2, 3, figsize=(18, 15))
axes = axes.flatten()

print("Scatter Plots of Features vs Price")

for i,features in enumerate(features):
    axes[i].scatter(df[features], df['Price'])
    axes[i].set_xlabel(features)
    axes[i].set_ylabel('Price')
    axes[i].set_title(f'{features} vs Price')
```

```python
# Using the Pearson Coefficient Method . Low value of correlation means that the two variables are not related to each other. A high value of correlation means that the two variables are related to each other. The value of correlation can be between -1 and 1. A value of 1 means that the two variables are perfectly correlated, a value of -1 means that the two variables are perfectly inversely correlated, and a value of 0 means that there is no correlation between the two variables.
numeric_df = df.select_dtypes(include='number')
numeric_df.corr(method='pearson')['Price'].sort_values(ascending=False)
```
